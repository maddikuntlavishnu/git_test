```python
df = sns.load_dataset("diamonds")
df.head()
```

```python
## cheking null values
df.isnull().sum()
```

```python
## separating the data set x and y
y = df['price']
x = df.drop('price',axis=1)
```

```python
x.info()
```

```python
x_num = x.select_dtypes(include ="float64")
x_num.head()
```

```python
x_char = x.select_dtypes(exclude="float64")
x_char.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for col in x_char:
    x_char[col] = le.fit_transform(x_char[col])
x_char.head()
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_scaling = pd.DataFrame(sc.fit_transform(x_char),
                         columns = x_char.columns,
                         index=x_char.index)
x_scaling.head()

                        
```

```python
data = pd.concat([x_char,x_scaling],axis=1)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    data,y, test_size=0.2,random_state=45)
```

```python
x_train.head()
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.neighbors import KNeighborsRegressor
neigh = 10
knn = KNeighborsRegressor(n_neighbors=neigh)
```

```python
knn.fit(x_train,y_train)
```

```python
ypred = knn.predict(x_test)
```

```python
## regression metrics
## mae,mse,rmse,r2
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
mae = mean_absolute_error(y_test,ypred)
print(mae)
mse = mean_squared_error(y_test,ypred)
print(mse)
rmse =root_mean_squared_error(y_test,ypred)
print(rmse)
r2 = r2_score(y_test,ypred)
r2
```
